## Weighting Rationale

The 50/50 Value-Momentum composite is selected as the primary specification because it provides a simple and interpretable baseline that assigns equal importance to two distinct signals.

The 70/30 and 30/70 specifications are included as robustness checks rather than being selected through optimization of historical returns. This avoids tuning the factor weights to maximize in-sample performance and provides a more defensible comparison of alternative weighting schemes.

The low correlation between Value and Momentum also supports combining the two signals. In the sample, the overall correlation between the standardized Value and Momentum signals is -0.0477, while the average monthly cross-sectional correlation is -0.0608. This suggests that the two signals provide relatively distinct information and may offer diversification benefits when combined.

## Task 1 Conclusion

The results indicate that combining Value and Momentum improves risk-adjusted performance relative to the individual factors.

The 50/50 composite achieves the highest Sharpe ratio among the tested specifications at 0.8185, compared with 0.7301 for Momentum and 0.4150 for Value. It also produces the highest annualized return at 14.16%, while maintaining lower annualized volatility of 17.30% and a smaller maximum drawdown of -21.76%.

The weighted alternatives also perform well, with Sharpe ratios of 0.8066 for the 70/30 Value-Momentum combination and 0.7939 for the 30/70 combination. However, neither improves upon the equal-weight specification.

The low correlation between Value and Momentum provides additional support for the combination. Overall, the 50/50 composite provides the strongest balance of return and risk among the tested specifications.

Because the 50/50 weighting was chosen as a simple, pre-specified equal-weight baseline rather than optimized to maximize in-sample performance, the result provides evidence that combining the two signals can improve risk-adjusted performance without relying on aggressive parameter fitting.

## Sector Neutralization Rationale

The sector-neutral portfolio is constructed by ranking stocks within each sector and selecting the top quintile within each sector. Each active sector is then assigned equal portfolio weight, with stocks within each sector equally weighted.

This approach is designed to reduce unintended sector concentration while preserving the relative factor ranking within each sector. The raw composite portfolio is retained as the baseline so that the effect of neutralization can be evaluated directly rather than assuming that neutralization is automatically beneficial.

The raw portfolio exhibits meaningful sector tilts, particularly toward Technology, Industrials, Financial Services, Healthcare, and Consumer Cyclical stocks. Sector neutralization reduces these exposures to approximately 9.09% per sector and lowers the sector HHI from 0.1637 to 0.0909.

## Task 2 Conclusion

Sector neutralization successfully removes the sector concentration present in the raw composite portfolio. The raw portfolio has an average Technology exposure of 21.57% and an average sector HHI of 0.1637, while the sector-neutral portfolio assigns approximately 9.09% to each of the 11 sectors and reduces HHI to 0.0909.

However, neutralization comes with a performance trade-off in this sample. Annualized volatility decreases from 17.30% to 16.36%, indicating that controlling sector exposure reduces portfolio risk. At the same time, annualized return declines from 14.16% to 12.37%, the Sharpe ratio falls from 0.8185 to 0.7563, and maximum drawdown increases from -21.76% to -23.68%.

Therefore, sector neutralization does not improve the overall risk-adjusted performance of the strategy in this sample. Instead, it reduces sector concentration and volatility at the cost of lower returns and a lower Sharpe ratio. This suggests that some of the raw composite's performance may be associated with sector tilts. Nevertheless, the neutralized version provides a useful robustness check and a more controlled measure of the underlying factor signal.

Sector neutralization does not improve the factor-adjusted performance in this sample, as both the estimated alpha and its statistical significance decline after neutralization.

## Task 4 Conclusion

The robustness analysis shows that the Raw Composite strategy generates a high absolute return of 28.32% annualized, but this performance is accompanied by substantial risk, including 66.17% annualized volatility and a maximum drawdown of -75.06%.

Performance is highly regime-dependent. The strategy performs strongly during bull markets, with an annualized return of 89.15% and a Sharpe ratio of 1.59, but performs poorly during bear markets, with an annualized return of -40.12% and a Sharpe ratio of -0.50. The 2021–2022 period represents the clearest failure period, with a -10.59% annualized return and a -75.06% maximum drawdown.

The quintile decomposition provides limited evidence of strong cross-sectional ranking ability. Q5 generates an average monthly return of 1.46%, compared with 1.42% for Q1, resulting in only a 0.04% monthly Q5-Q1 spread. This suggests that a substantial portion of the strategy's historical performance may be driven by systematic exposures rather than a strong monotonic composite signal.

The signal shows some persistence over longer horizons, with average forward returns increasing from 1.46% at one month to 17.29% at twelve months. Parameter sensitivity is moderate: the strategy remains profitable across different momentum lookback and skip-period specifications, although performance varies across specifications.

The strategy also appears robust to restricting the universe to large-cap stocks. The large-cap universe produces a higher Sharpe ratio of 0.64 compared with 0.43 for the full universe, suggesting that the strategy does not depend primarily on small-cap stocks. Liquidity-weighted implementation also improves the Sharpe ratio to 0.60 from 0.43 for equal weighting.

Overall, the robustness analysis indicates that the strategy has attractive absolute return potential but remains highly exposed to market regimes and substantial drawdown risk. The weak Q5-Q1 spread and strong dependence on bull-market conditions suggest that its historical performance should not be interpreted as pure stock-selection alpha. The results support further investigation of alternative weighting and implementation choices, particularly liquidity weighting and broader portfolio selection, before selecting the final strategy.

# Task 5: Research Synthesis Memo

```python
import pandas as pd
from pathlib import Path


print("TASK 5: RESEARCH SYNTHESIS")



# 1. LOAD SAVED RESULTS


BASE_DIR = Path.cwd()

task1_perf = pd.read_csv(
    BASE_DIR / "task1_performance_comparison.csv",
    index_col=0
)

task2_perf = pd.read_csv(
    BASE_DIR / "task2_performance_comparison.csv",
    index_col=0
)

task3_attr = pd.read_csv(
    BASE_DIR / "task3_attribution_summary.csv",
    index_col=0
)

print("\nTask 1 performance:")
print(task1_perf.round(4))

print("\nTask 2 performance:")
print(task2_perf.round(4))

print("\nTask 3 attribution:")
print(task3_attr.round(4))



# 2. TASK 1: FACTOR COMBINATION COMPARISON



print("TASK 1: FACTOR COMBINATION")


# Keep the relevant strategies if they exist
task1_candidates = [
    "Value",
    "Momentum",
    "Composite_50_50",
    "Composite_70_30",
    "Composite_30_70"
]

available_candidates = [
    x for x in task1_candidates
    if x in task1_perf.index
]

task1_candidates_table = task1_perf.loc[
    available_candidates
].copy()

print(task1_candidates_table.round(4))



# 3. IDENTIFY BEST TASK 1 COMPOSITE


composite_candidates = [
    x for x in [
        "Composite_50_50",
        "Composite_70_30",
        "Composite_30_70"
    ]
    if x in task1_perf.index
]

best_composite = (
    task1_perf.loc[composite_candidates, "Sharpe"]
    .idxmax()
)

print("\nBest composite based on Sharpe:")
print(best_composite)

print(
    "Sharpe:",
    round(task1_perf.loc[best_composite, "Sharpe"], 4)
)



# 4. TASK 2: RAW VS SECTOR-NEUTRAL



print("TASK 2: RAW VS SECTOR-NEUTRAL")


print(task2_perf.round(4))



# 5. TASK 3: ATTRIBUTION



print("TASK 3: FOUR-FACTOR ATTRIBUTION")


print(task3_attr.round(4))



# 6. FINAL CANDIDATE COMPARISON



print("FINAL CANDIDATE COMPARISON")


# Build a compact comparison table from the saved results

comparison_rows = []

# Task 1 candidates
for strategy in available_candidates:
    row = {
        "Strategy": strategy,
        "Source": "Task 1",
        "Annualized Return": task1_perf.loc[
            strategy, "Annualized Return"
        ],
        "Annualized Volatility": task1_perf.loc[
            strategy, "Annualized Volatility"
        ],
        "Sharpe": task1_perf.loc[
            strategy, "Sharpe"
        ],
        "Max Drawdown": task1_perf.loc[
            strategy, "Max Drawdown"
        ]
    }

    comparison_rows.append(row)


# Task 2 candidates
for strategy in task2_perf.index:
    row = {
        "Strategy": strategy,
        "Source": "Task 2",
        "Annualized Return": task2_perf.loc[
            strategy, "Annualized Return"
        ],
        "Annualized Volatility": task2_perf.loc[
            strategy, "Annualized Volatility"
        ],
        "Sharpe": task2_perf.loc[
            strategy, "Sharpe"
        ],
        "Max Drawdown": task2_perf.loc[
            strategy, "Max Drawdown"
        ]
    }

    comparison_rows.append(row)


final_comparison = pd.DataFrame(comparison_rows)

print(final_comparison.round(4))

final_comparison.to_csv(
    BASE_DIR / "task5_candidate_comparison.csv",
    index=False
)

print("\nTask 5 candidate comparison saved.")
```
